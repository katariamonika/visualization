```python
"""
Visualization— Live Learning Session 1
Date: 2026-01-15

Overview:
- Learned how to create a basic figure using matplotlib.
- Generated small sample data using NumPy.
- Created a simple scatter plot and confirmed it renders correctly.

Notes / takeaways:
- Using a fixed random seed helps make plots reproducible (same random values each run).
- plt.subplots() is a clean way to create a Figure + Axes object for plotting.
- Scatter plots are useful for showing relationships/patterns between two numeric variables.
"""
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# Create sample data
np.random.seed(613)

x = np.arange(50)
y = np.random.randint(0, 100, 50)
```

```python
# Basic scatter plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y)
```

```python
# Change plot type
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x, y)
```

```python
# Histogram
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)

```

```python
# Histogram with edge color
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y, edgecolor="black")
```

```python
# Line plot with labels and title

fig, ax = plt.subplots(figsize=(5, 3))

ax.plot(x, y)

ax.set_title("Total growth over time")
ax.set_ylabel("Total growth")
ax.set_xlabel("Years since start")

fig.tight_layout()
```

```python
# Custom Fonts for Labels and Title
font1 = {'family': 'sans-serif', 'color': 'blue', 'size': 20}
font2 = {'family': 'monospace', 'color': 'green', 'size': 14}

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y)

ax.set_title("Total growth over time", fontdict=font1)
ax.set_ylabel("Total growth", fontdict=font2)
ax.set_xlabel("Years since start", fontdict=font2)

fig.tight_layout()
plt.show()

```

```python
# Styled Scatter Plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y, marker="*", color="indigo")
plt.show()
```

```python
# Styled Line Plot (Markers, Linestyle, Width)
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(
    x,
    y,
    marker="*",
    color="indigo",
    linestyle="--",
    linewidth=2
)
plt.show()
```

```python
#Further Customization (Markers and Colors)

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(
    x,
    y,
    marker="*",
    markersize=12,
    linestyle="--",
    linewidth=2,
    color="#7425b9",
    markeredgecolor="#f1a359",
    markerfacecolor="#000000"
)
plt.show()
```

```python
#Add Grid Lines
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(
    x,
    y,
    marker="*",
    markersize=12,
    linestyle="--",
    linewidth=2
)

ax.grid(axis="y")
plt.show()
```

```python
# Styled Grid Lines
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(
    x,
    y,
    marker="*",
    markersize=12,
    linestyle="--",
    linewidth=2
)

ax.grid(axis="y", color="blue", linewidth=2, linestyle=":")
plt.show()
```
